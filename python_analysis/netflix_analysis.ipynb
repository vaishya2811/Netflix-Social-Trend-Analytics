```python
netflix_df = netflix_df.drop_duplicates()
```

```python
netflix_df.shape
```

```python
netflix_df['country'] = netflix_df['country'].fillna('Unknown')
netflix_df['director'] = netflix_df['director'].fillna('Not Available')
netflix_df['cast'] = netflix_df['cast'].fillna('Not Available')
netflix_df['rating'] = netflix_df['rating'].fillna('Not Rated')
netflix_df['date_added'] = netflix_df['date_added'].fillna('Unknown')
netflix_df['duration'] = netflix_df['duration'].fillna('Unknown')
```

```python
netflix_df.isnull().sum()
```

```python
netflix_df.dtypes
```

```python
netflix_df['date_added'] = pd.to_datetime(netflix_df['date_added'], errors='coerce')
netflix_df.info()
netflix_df['date_added'].isnull().sum()
```

```python
netflix_df['year_added'] = netflix_df['date_added'].dt.year
netflix_df[['title', 'year_added']].head()
```

```python
yearly_content = netflix_df['year_added'].value_counts().sort_index()

yearly_content
plt.figure(figsize=(12,6))

yearly_content.plot(kind='line', marker='o')

plt.title('Netflix Content Added Over Years')
plt.xlabel('Year')
plt.ylabel('Number of Titles Added')

plt.show()
```

```python
netflix_df['type'].value_counts()
plt.figure(figsize=(6,6))

netflix_df['type'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title('Movies vs TV Shows on Netflix')

plt.ylabel('')

plt.show()
```

```python
genres = netflix_df['listed_in'].str.split(', ').explode()

top_genres = genres.value_counts().head(10)

top_genres
plt.figure(figsize=(12,6))

top_genres.plot(kind='bar')

plt.title('Top Genres on Netflix')
plt.xlabel('Genre')
plt.ylabel('Count')

plt.xticks(rotation=45)

plt.show()
```

```python
top_countries = netflix_df['country'].value_counts().head(10)

top_countries
plt.figure(figsize=(12,6))

top_countries.plot(kind='bar')

plt.title('Top Content Producing Countries')
plt.xlabel('Country')
plt.ylabel('Number of Titles')

plt.xticks(rotation=45)

plt.show()
```

```python
imdb_df.head()



imdb_df.info()


imdb_df.isnull().sum()
```

```python
imdb_df = imdb_df.drop_duplicates()
imdb_df['Gross'] = imdb_df['Gross'].fillna('Unknown')
imdb_df['Meta_score'] = imdb_df['Meta_score'].fillna(imdb_df['Meta_score'].median())
imdb_df['Certificate'] = imdb_df['Certificate'].fillna('Not Rated')
imdb_df.isnull().sum()
```

```python
genre_ratings = imdb_df.groupby('Genre')['IMDB_Rating'].mean().sort_values(ascending=False).head(10)

genre_ratings

plt.figure(figsize=(12,6))

genre_ratings.plot(kind='bar')

plt.title('Top Rated Genres by IMDb Rating')

plt.xlabel('Genre')

plt.ylabel('Average IMDb Rating')

plt.xticks(rotation=45)

plt.show()
```

TOP RATED GENRES

```python
top_voted = imdb_df.sort_values(by='No_of_Votes', ascending=False).head(10)

top_voted[['Series_Title', 'No_of_Votes']]

plt.figure(figsize=(12,6))

plt.bar(top_voted['Series_Title'], top_voted['No_of_Votes'])

plt.title('Most Popular Movies by Audience Votes')

plt.xlabel('Movie')

plt.ylabel('Votes')

plt.xticks(rotation=75)

plt.show()
```

MOST VOTED MOVIES

```python
imdb_df['Released_Year'] = pd.to_numeric(imdb_df['Released_Year'], errors='coerce')


recent_movies = imdb_df['Released_Year'].value_counts().sort_index().tail(30)

recent_movies.plot(figsize=(12,6))

plt.title('Movie Release Trend Over Years')

plt.xlabel('Year')

plt.ylabel('Number of Movies')

plt.grid()

plt.show()
```
